```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
print("All libraries imported successfully!")
```

```python
df = pd.read_excel("BlinkIT Grocery Data.xlsx")
print("Dataset loaded successfully!")
print("Number of rows and columns:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())
print("\nFirst 5 rows of the dataset:")
display(df.head())
```

```python
print("DATA TYPES OF EACH COLUMN")
print("-" * 40)
print(df.dtypes)
print("\nDATASET INFORMATION")
print("-" * 40)
df.info()
print("\nSUMMARY STATISTICS")
print("-" * 40)
display(df.describe())
print("\nUNIQUE VALUES IN IMPORTANT COLUMNS")
print("-" * 40)
for col in ["Item Type", "Outlet Type", "Outlet Size", "Outlet Location Type"]:
    print(f"\n{col}:")
    print(df[col].unique())
```

```python
print("MISSING VALUES IN EACH COLUMN")
print("-" * 40)
display(df.isnull().sum())
print("\nTOTAL MISSING VALUES")
print("-" * 40)
print(df.isnull().sum().sum())
print("\nDUPLICATE ROWS")
print("-" * 40)
print("Number of duplicate rows:", df.duplicated().sum())
print("\nMISSING ITEM WEIGHT RECORDS")
print("-" * 40)
display(df[df["Item Weight"].isnull()].head(10))
```

```python

print("1. ITEM FAT CONTENT LABELS")
display(df["Item Fat Content"].value_counts())
print("\n2. OUTLET SIZE LABELS")
display(df["Outlet Size"].value_counts())
print("\n3. ITEM VISIBILITY SUMMARY")
display(df["Item Visibility"].describe())
print("\n4. ITEM WEIGHT SUMMARY")
display(df["Item Weight"].describe())
print("\n5. SALES SUMMARY")
display(df["Sales"].describe())
print("\n6. CUSTOMER RATING SUMMARY")
display(df["Rating"].describe())
```

```python
print("Item Fat Content:")
display(df["Item Fat Content"].value_counts())
print("\nOutlet Size:")
display(df["Outlet Size"].value_counts())
```

```python
df_clean = df.copy()
df_clean["Item Fat Content"] = (
    df_clean["Item Fat Content"]
    .replace({
        "LF": "Low Fat",
        "low fat": "Low Fat",
        "reg": "Regular"
    })
)
print("Item Fat Content after cleaning:")
display(df_clean["Item Fat Content"].value_counts())
print("\nOutlet Size after cleaning:")
display(df_clean["Outlet Size"].value_counts())
```

```python
weight_check = (
    df_clean.groupby("Item Identifier")["Item Weight"]
    .agg(["count", "first"])
    )
missing_weight_products = df_clean.loc[
    df_clean["Item Weight"].isnull(),
    "Item Identifier"
].unique()
known_weight_products = df_clean.loc[
    df_clean["Item Weight"].notnull(),
    "Item Identifier"
].unique()
print("Records with missing Item Weight:",
      df_clean["Item Weight"].isnull().sum())
print("Unique products with missing Item Weight:",
      len(missing_weight_products))
print("Unique products with known Item Weight:",
      len(known_weight_products))
print("\nExample records with missing weights:")
display(
    df_clean.loc[
        df_clean["Item Weight"].isnull(),
        ["Item Identifier", "Item Type", "Item Weight"]
    ].head(10)
)
```

```python
known_weights = (
    df_clean.dropna(subset=["Item Weight"])
    .groupby("Item Identifier")["Item Weight"]
    .first()
)
df_clean["Item Weight"] = df_clean["Item Weight"].fillna(df_clean["Item Identifier"].map(known_weights))
print("Missing Item Weight values after matching products:")
print(df_clean["Item Weight"].isnull().sum())
print("\nRemaining missing weights by product type:")
display(df_clean[df_clean["Item Weight"].isnull()].groupby("Item Type").size().sort_values(ascending=False))
```

```python
category_medians = df_clean.groupby("Item Type")["Item Weight"].median()
missing_mask = df_clean["Item Weight"].isnull()
df_clean.loc[missing_mask, "Item Weight"] = (df_clean.loc[missing_mask, "Item Type"].map(category_medians))
print("Missing values after final cleaning:")
display(df_clean.isnull().sum())
print("\nTotal missing values remaining:",df_clean.isnull().sum().sum())
```

```python
print("BLINKIT GROCERY SALES ANALYSIS")
print("=" * 40)
total_sales = df_clean["Sales"].sum()
average_sales = df_clean["Sales"].mean()
average_rating = df_clean["Rating"].mean()
unique_products = df_clean["Item Identifier"].nunique()
unique_outlets = df_clean["Outlet Identifier"].nunique()
print(f"Total Sales: {total_sales:,.2f}")
print(f"Average Sales per Record: {average_sales:,.2f}")
print(f"Average Customer Rating: {average_rating:.2f} / 5")
print(f"Number of Unique Products: {unique_products}")
print(f"Number of Unique Outlets: {unique_outlets}")
print(f"Total Sales Records: {len(df_clean)}")
```

```python
category_sales = (df_clean.groupby("Item Type")["Sales"].sum().sort_values(ascending=False))
print("TOTAL SALES BY PRODUCT CATEGORY")
display(category_sales.reset_index())
plt.figure(figsize=(12, 6))
category_sales.sort_values().plot(kind="barh")
plt.title("BlinkIT Sales by Product Category")
plt.xlabel("Total Sales")
plt.ylabel("Product Category")
plt.tight_layout()
plt.show()
print("\nHighest-selling category:",category_sales.idxmax())
print("Sales value:",round(category_sales.max(), 2))
```

```python
outlet_sales = (df_clean.groupby("Outlet Type")["Sales"].sum().sort_values(ascending=False))
print("TOTAL SALES BY OUTLET TYPE")
display(outlet_sales.reset_index())
plt.figure(figsize=(10, 5))
outlet_sales.plot(kind="bar")
plt.title("BlinkIT Sales by Outlet Type")
plt.xlabel("Outlet Type")
plt.ylabel("Total Sales")
plt.xticks(rotation=30, ha="right")
plt.tight_layout()
plt.show()
print("Highest-performing outlet type:", outlet_sales.idxmax())
print("Total sales:", round(outlet_sales.max(), 2))
```

```python
location_sales = (
     df_clean.groupby("Outlet Location Type")["Sales"].sum().sort_values(ascending=False)
     )
print("TOTAL SALES BY OUTLET LOCATION")
display(location_sales.reset_index())
plt.figure(figsize=(8, 5))
location_sales.plot(kind="bar")
plt.title("BlinkIT Sales by Outlet Location")
plt.xlabel("Outlet Location")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
print("Highest-performing location:", location_sales.idxmax())
print("Total sales:", round(location_sales.max(), 2))
```

```python

outlet_ratings = (
    df_clean.groupby("Outlet Type")["Rating"].mean().sort_values(ascending=False)
    )
print("AVERAGE CUSTOMER RATING BY OUTLET TYPE")
display(outlet_ratings.reset_index())
plt.figure(figsize=(10, 5))
outlet_ratings.plot(kind="bar")
plt.title("Average Customer Rating by Outlet Type")
plt.xlabel("Outlet Type")
plt.ylabel("Average Rating")
plt.ylim(0, 5)
plt.xticks(rotation=30, ha="right")
plt.tight_layout()
print("Highest-rated outlet type:", outlet_ratings.idxmax())
print("Average rating:", round(outlet_ratings.max(), 2))
```

```python
fat_sales = (
    df_clean.groupby("Item Fat Content")["Sales"].sum().sort_values(ascending=False)
)

print("TOTAL SALES BY ITEM FAT CONTENT")
display(fat_sales.reset_index())
plt.figure(figsize=(7, 5))
fat_sales.plot(kind="bar")
plt.title("Total Sales by Item Fat Content")
plt.xlabel("Item Fat Content")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
print("Highest-selling fat-content group:", fat_sales.idxmax())
print("Total sales:", round(fat_sales.max(), 2))
```

```python
numeric_columns = [
    "Item Weight",
    "Item Visibility",
    "Sales",
    "Rating"
]
plt.figure(figsize=(12, 8))
for i, column in enumerate(numeric_columns, 1):
    plt.subplot(2, 2, i)
    sns.boxplot(y=df_clean[column])
    plt.title(f"Boxplot of {column}")
plt.tight_layout()
plt.show()
print("POTENTIAL OUTLIER COUNTS")
print("-" * 35)
for column in numeric_columns:
    Q1 = df_clean[column].quantile(0.25)
    Q3 = df_clean[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR
count = (
        (df_clean[column] < lower_limit) |
        (df_clean[column] > upper_limit)
    ).sum()
print(f"{column}: {count} potential outliers")
```
